data cleaning

```python
df['Order Date']=pd.to_datetime(df['Order Date'],format='%d/%m/%Y')
df['Ship Date']=pd.to_datetime(df['Ship Date'],format='%d/%m/%Y')
df.info()
```

```python
df['Postal Code'] = df['Postal Code'].fillna(df.groupby('City')['Postal Code'].transform('median'))
print(df['Postal Code'])
```

```python
df.isnull().sum()
```

```python
# Convert Postal Code to string/integer representation
df['Postal Code'] = df['Postal Code'].astype(int).astype(str).str.zfill(5)
# Convert repetitive object columns to category
categorical_cols = ['Ship Mode', 'Segment', 'Country', 'Region', 'Category', 'Sub-Category']
df[categorical_cols] = df[categorical_cols].astype('category')
```

```python
# Calculate shipping delay in days
df['Shipping Days'] = (df['Ship Date'] - df['Order Date']).dt.days
df['Shipping Days']
```

```python
# Extract time components for time-series analysis
df['Order Year'] = df['Order Date'].dt.year
df['Order Month'] = df['Order Date'].dt.month
df['Order DayOfWeek'] = df['Order Date'].dt.day_name()
```

```python
text_cols = ['Customer Name', 'City', 'State', 'Product Name']
for col in text_cols:
    df[col] = df[col].astype(str).str.strip()
```

```python
df[df['Shipping Days'] < 0]
```

Exploratory Data Analysis

```python
cols_to_drop = ['Row ID', 'Country']
df = df.drop(columns=cols_to_drop)
```

```python
print(df['Sales'].describe())
```

```python
# Grouping states with very few orders into 'Other'
state_counts = df['State'].value_counts()
top_states = state_counts[state_counts >= 50].index
df['State_Grouped'] = df['State'].apply(lambda x: x if x in top_states else 'Other')
```

```python
# Total sales by Category
print(df.groupby('Category', observed=False)['Sales'].sum())

# Average shipping delay by Ship Mode
print(df.groupby('Ship Mode', observed=False)['Shipping Days'].mean())
```

```python
plt.figure(figsize=(12, 4))

# Sales Distribution (Log Scale due to right skew)
plt.subplot(1, 2, 1)
sns.histplot(df['Sales'], bins=50, kde=True, log_scale=True, color='teal')
plt.title('Log-Scaled Sales Distribution')

# Shipping Days Distribution
plt.subplot(1, 2, 2)
sns.countplot(data=df, x='Shipping Days',hue='Shipping Days', palette='viridis')
plt.title('Distribution of Shipping Days')

plt.tight_layout()
plt.show()
```

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Total Sales by Region (Updated parameters)
sns.barplot(data=df, x='Region', y='Sales', estimator=sum, errorbar=None, hue='Region', ax=axes[0], palette='Blues_d', legend=False)
axes[0].set_title('Total Sales by Region')
axes[0].set_ylabel('Total Sales ($)')

# Total Sales by Sub-Category (Updated parameters)
sub_cat_sales = df.groupby('Sub-Category', observed=False)['Sales'].sum().sort_values(ascending=False)
sns.barplot(x=sub_cat_sales.values, y=sub_cat_sales.index, hue=sub_cat_sales.index, ax=axes[1], palette='crest', legend=False)
axes[1].set_title('Total Sales by Sub-Category')
axes[1].set_xlabel('Total Sales ($)')

plt.tight_layout()
plt.show()
```

```python
top_states_sales = df.groupby('State_Grouped', observed=False)['Sales'].sum().sort_values(ascending=False)

plt.figure(figsize=(10, 5))
top_states_sales.plot(kind='bar', color='darkblue')
plt.title('Total Revenue by Top States (Grouped)')
plt.ylabel('Sales ($)')
plt.xlabel('State')
plt.xticks(rotation=45)
plt.show()
```

```python
# Aggregate monthly sales across years
monthly_sales = df.groupby(['Order Year', 'Order Month'], observed=False)['Sales'].sum().reset_index()

plt.figure(figsize=(12, 5))
sns.lineplot(data=monthly_sales, x='Order Month', y='Sales', hue='Order Year', marker='o', palette='tab10')
plt.title('Monthly Revenue Performance by Year')
plt.xlabel('Month')
plt.ylabel('Total Sales ($)')
plt.xticks(range(1, 13))
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
```

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Sales distribution across Customer Segments
sns.barplot(data=df, x='Segment', y='Sales', estimator=sum, errorbar=None, hue='Segment', ax=axes[0], palette='Set2', legend=False)
axes[0].set_title('Total Revenue by Customer Segment')

# Average Order Value (AOV) by Ship Mode (Fixed by using 'mean' or np.mean)
sns.barplot(data=df, x='Ship Mode', y='Sales', estimator='mean', errorbar=None, hue='Ship Mode', ax=axes[1], palette='Pastel1', legend=False)
axes[1].set_title('Average Order Value (AOV) by Ship Mode')

plt.tight_layout()
plt.show()
```
